```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import glob
import re
```

```python
y_scale = "y2"
y_scale_num = 2

# Define the base directory, output name, and the categories
base_dir = '/puffin_outputs'
outfilename = 'Supplementary1'

categories = {
    'HPRT1': 1,
    'Chr8 Holdout': 1
}

titles = [
    'FANTOM_CAGE (forward strand)',
    'ENCODE_CAGE (forward strand)',
    'ENCODE_RAMPAGE (forward strand)',
    'GRO_CAP (forward strand)',
    'PRO_CAP (forward strand)',
    'PRO_CAP (reverse strand)',
    'GRO_CAP (reverse strand)',
    'ENCODE_RAMPAGE (reverse strand)',
    'ENCODE_CAGE (reverse strand)',
    'FANTOM_CAGE (reverse strand)'
]

# Function to load numpy arrays from a given directory
def load_numpy_arrays(category, num_files):
    arrays = []
    filenames = []
    category_dir = os.path.join(base_dir, category)
    for npy_file in glob.glob(os.path.join(category_dir, '*.npy'))[:num_files]:
        arrays.append(np.load(npy_file))
        filenames.append(os.path.basename(npy_file))
    return arrays, filenames

# Function to plot a single numpy array
def plot_array(array, title, output_dir, base_filename):
    fig, axes = plt.subplots(5, 2, figsize=(15, 20))  # 5 rows and 2 columns for 10 plots
    axes = axes.flatten()
    
    # Rearrange the default Puffin-D assay order
    plot_order = [0, 9, 1, 8, 2, 7, 3, 6, 4, 5]
    
    for i in range(array.shape[1]):
        idx = plot_order[i]
        axes[i].plot(array[0, idx, :])
        axes[i].set_title(titles[idx])
        axes[i].set_xlabel('Position')
        axes[i].set_ylabel('Value')
        axes[i].set_xlim([0, 100000])
        axes[i].set_ylim([0, y_scale_num])
    
    fig.suptitle(title, fontsize=16)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])  # Adjust layout to make room for the suptitle
    plot_filename = f'{title}_{base_filename}_plot_{y_scale}.png'
    plt.savefig(os.path.join(output_dir, plot_filename))
    plt.close()
    
```

```python
# Run functions and plot

if __name__ == '__main__':
    
    all_data = {} #Create a dictionary to hold all the data
    
    for category, num_files in categories.items():
        # Load the numpy arrays and their filenames
        arrays, filenames = load_numpy_arrays(category, num_files)
        
        # Store data in the dictionary
        all_data[category] = {'arrays': arrays, 'filenames': filenames}
        
        # Create an output directory for the plots if it doesn't exist
        output_dir = os.path.join('/plots', category, y_scale)
        os.makedirs(output_dir, exist_ok=True)
        
        # Plot each array
        for idx, (array, filename) in enumerate(zip(arrays, filenames)):
            plot_title = f'{category} File {idx + 1}'
            base_filename = os.path.splitext(filename)[0]  # Get the base filename without extension
            plot_array(array, category, output_dir, base_filename)
            print(f'Plots for {category} completed.')
            
            
            
```
